# Funtions 

```python
def get_stats(api_key, channel_id):

    url = (
        "https://youtube.googleapis.com/youtube/v3/channels"
        f"?part=snippet,statistics&id={channel_id}&key={api_key}"
    )

    response = requests.get(url).json()

    item = response["items"][0]
    stats = item["statistics"]

    date = datetime.now().strftime("%Y-%m-%d")

    data_channel = {
        "Channel_id": channel_id,
        "Created_at": date,
        "Channel_name": item["snippet"]["title"],
        "Total_Views": int(stats["viewCount"]),
        "Subscribers": int(stats["subscriberCount"]),
        "Video_count": int(stats["videoCount"])
  }
    
    return data_channel
```

# Channels to analize

```python

with open("channels.json", "r", encoding="utf-8") as f:
        channel_ids = json.load(f)
```

```python
def channel_stats(channel_ids, api_key):

    data = []

    tiempo = [1, 2.5, 3, 2]

    for channel_id in tqdm(channel_ids, colour="green"):

        stats = get_stats(api_key, channel_id)
        data.append(stats)

        time.sleep(random.choice(tiempo))

    return pd.DataFrame(data)
```

```python
df_channels_final = channel_stats(channel_ids, DEVELOPER_KEY)
df_channels_final
```

```python
df_channels_short = df_channels_final[
    ["Channel_name", "Channel_id"]
].copy()

df_channels_short.to_csv("channels_to_analize.csv", index=False)
df_channels_short
```
